```python
import os
import subprocess
from pathlib import Path

# ====== CONFIGURATION ======

INPUT_ROOT = Path("./data")
OUTPUT_ROOT = Path("./data_processed")

# Target settings
TARGET_WIDTH = 640     # downscale width
TARGET_HEIGHT = 480    # downscale height
TARGET_FPS = 15        # reduce framerate

CRF_QUALITY = 23       # 18=high quality, 23=default, 28=smaller file

print("Input directory:", INPUT_ROOT.resolve())
print("Output directory:", OUTPUT_ROOT.resolve())
```

```python
def process_video(input_path, output_path):
    output_path.parent.mkdir(parents=True, exist_ok=True)

    command = [
        "ffmpeg",
        "-y",  # overwrite existing files
        "-i", str(input_path),
        "-vf", f"scale={TARGET_WIDTH}:{TARGET_HEIGHT},fps={TARGET_FPS}",
        "-c:v", "libx264",
        "-preset", "medium",
        "-crf", str(CRF_QUALITY),
        "-pix_fmt", "yuv420p",
        "-an",  # remove audio (usually unnecessary for ML)
        str(output_path)
    ]

    print(f"Processing: {input_path.name}")
    subprocess.run(command)
    print(f"Saved → {output_path}")
```

```python
def batch_process_videos():
    for category_dir in INPUT_ROOT.iterdir():
        if category_dir.is_dir():
            category_name = category_dir.name
            for video_file in category_dir.glob("*.mov"):
                output_dir = OUTPUT_ROOT / category_name
                output_file = output_dir / (video_file.stem + ".mp4")

                process_video(video_file, output_file)


    print("\n All videos processed.")

batch_process_videos()
```
